# **Exercise Questions On Sets:**

An **unordered** collection of **unique** values. Values within {curly brackets} and comma-separated.

* Duplicates are removed automatically.
* Sets have no order, so there is no indexing (no `s[0]`).
* Sets are mutable: you can add and remove items.

```python
colours = {'red', 'green', 'blue', 'red'}
print(colours)          # 'red' appears only once
print(type(colours))
```

```python
colours[0]              # TypeError: sets have no positions, so no indexing
```

**Q2: The sets below show students enrolled in two modules. Write code to find:**

* Students taking **both** modules
* Students taking **either** module
* Students taking **only** CMP7005
* Students taking **only one** of the two modules

**Given:**

**cmp7005** = {'Ali', 'Ben', 'Chloe', 'Dev'}

**dataviz** = {'Chloe', 'Dev', 'Ema', 'Farah'}

```python
# Solution to Q2:
cmp7005 = {'Ali', 'Ben', 'Chloe', 'Dev'}
dataviz = {'Chloe', 'Dev', 'Ema', 'Farah'}

print("Both:", cmp7005 & dataviz)            # intersection
```

```python
print("Either:", cmp7005 | dataviz)          # union
```

```python
print("Only CMP7005:", cmp7005 - dataviz)    # difference
```

```python
print("Only one:", cmp7005 ^ dataviz) # symmetric difference
```

**Q3: Merge two Python dictionaries into one**

**Write a code to merge two dictionaries into a new dictionary and print it.**

**Given:**

**dict1** = {'Ten': 10, 'Twenty': 20, 'Thirty': 30}

**dict2** = {'Thirty': 30, 'Forty': 40, 'Fifty': 50}

```python
# Solution to Q3:
# Dictionary unpacking is a feature that allows you to "unpack" the key-value pairs from a dictionary into another dictionary or function arguments. It uses the ** operator.

dict1 = {'Ten': 10, 'Twenty': 20, 'Thirty': 30}
dict2 = { 'Forty': 40, 'Fifty': 50, 'Sixty': 60}

dict3 = {**dict1, **dict2}
print(dict3)
```

```python
dict4 = {**dict2, **dict1}
print("Changing the merge order:", dict4)   # 'Thirty': 30
```

**Q4: Given a string, create a dictionary where keys are characters and values are their frequencies in the string.**

**Given:**

**string1 = 'Jessa'**

**Expected Output :**

**Frequencies for 'Jessa': {'J': 1, 'e': 1, 's': 2, 'a': 1}**

```python
# Solution to Q4:
from collections import Counter

string1 = 'Jessa'
freq_dict = Counter(string1)

print(dict(freq_dict))
```

```python
# Alternatively:

string1 = 'Jessa'

freq_dict = {char: string1.count(char) for char in string1}
print(freq_dict)
```

**Q5: Initialise a dictionary where every subject starts with a mark of 0. Then change the mark for 'maths' to 75 and print the dictionary.**

**Given:**

**subjects** = ['maths', 'english', 'science']

**Expected output:**

{'maths': 0, 'english': 0, 'science': 0}

{'maths': 75, 'english': 0, 'science': 0}

```python
# Solution to Q5:
subjects = ['maths', 'english', 'science']

marks = dict.fromkeys(subjects, 0)
print(marks)
```

```python
# Solution to Q5:

marks['maths'] = 75
print(marks)

# fromkeys() gives every key the same starting value.
# This is safe here because 0 is a number, and numbers cannot be changed in place.
```

**Q6 (Stretch task): The code below uses `dict.fromkeys()` to give two employees the same default details. Give Kelly a salary of 9000 and then print Emma's details. What happened to Emma? Can you explain why, and fix the code so only Kelly's salary changes?**

**Given:**

**employees** = ['Kelly', 'Emma']

**defaults** = {"designation": 'Developer', "salary": 8000}

**Expected output (after fixing):**

{'Kelly': {'designation': 'Developer', 'salary': 9000}, 'Emma': {'designation': 'Developer', 'salary': 8000}}

```python
# Solution to Q6:
employees = ['Kelly', 'Emma']
defaults = {"designation": 'Developer', "salary": 8000}

res = dict.fromkeys(employees, defaults)
res['Kelly']['salary'] = 9000
#print(res['Emma'])   # Emma's salary is also 9000!
print(res)
```

```python
# Both Kelly and Emma point to the SAME defaults dictionary.
# Note: the line above also changed 'defaults' itself, so reset it first.
# Fix: give each employee their own copy.

defaults = {"designation": 'Developer', "salary": 8000}
res = {'Kelly': defaults.copy(), 'Emma': defaults.copy()}
res['Kelly']['salary'] = 9000
print(res)
```

**Key point:** `=` gives a second name to the same dictionary. `.copy()` creates a new, separate dictionary.
